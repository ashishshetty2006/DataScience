```python
import pandas as pd
import numpy as np
data={
    'Employee':['John','John','David','David','Alice','Alice','John','David'],
    'Department':['IT','HR','IT','HR','IT','HR','Finance','Finance'],
    'City':['Chennai','Chennai','Bangalore','Bangalore','Chennai','Delhi','Delhi','Delhi'],
    'Salary':[50000,45000,60000,55000,70000,65000,48000,62000],
    'Bonus':[5000,3000,7000,6000,8000,7500,3500,6500]
}
df=pd.DataFrame(data)
print(df)
```

```python
pivot=pd.pivot_table(df,values='Salary',index='Department')
print(pivot)
```

```python
pivot=pd.pivot_table(df,values='Salary',index='Department',aggfunc='sum')
print(pivot)

```

```python
pivot=pd.pivot_table(df,values='Salary',index='Department',aggfunc='min')
print(pivot)
```

```python
pivot=pd.pivot_table(df,values='Salary',index='Department',aggfunc='max')
print(pivot)
```

```python
pivot=pd.pivot_table(df,values='Salary',index='Department',aggfunc='count')
print(pivot)
```

```python
pivot=pd.pivot_table(df,values='Salary',index='Department',aggfunc='mean')
print(pivot)
```

```python
pivot=pd.pivot_table(df,values='Salary',index='Department',aggfunc=['sum','count','mean','max','min'])
print(pivot)
```

```python
pivot=pd.pivot_table(df,values='Salary',index='Department',columns='City',aggfunc='sum')
print(pivot)
```

```python
pivot=pd.pivot_table(df,values='Salary',index='Department',columns='City',aggfunc='sum',fill_value=40000)
print(pivot)
```

```python
pivot=pd.pivot_table(df,values='Salary',index=['Department','Employee'],aggfunc='sum')
print(pivot)
```

```python
pivot=pd.pivot_table(df,values=['Salary','Bonus'],index='Department',aggfunc=['sum','mean'])
print(pivot)                   
```

```python
pivot=pd.pivot_table(df,values='Salary',index='Department',aggfunc='sum',margins=True)
print(pivot)                   
```

```python
pivot=pd.pivot_table(df,values=['Salary','Bonus'],index='Department',aggfunc={'Salary':'sum','Bonus':'mean'})
print(pivot)                   
```

```python
dept_salary=df.groupby('Department')['Salary'].sum()
print(dept_salary)
```

```python
dept_salary_mean=df.groupby('Department')['Salary'].mean()
print(dept_salary_mean)
```
